```python
import pandas as pd
df =pd.read_csv('cleaned_covid19_data.csv')
print(df.to_string()) 
```

```python
# TASK 1-----------------------------------------------------------------
#Countries by Total Cases
#Calculation: Sum Cases per Country, sort descending, take top 10.
```

```python
#first calculation-------------------------------------------------------
```

```python
cal = df.groupby('Country')['Cases'].sum().sort_values(ascending = False).head(10)
cal
```

```python
#visulization for cal----------------------------------------------------
```

```python
import matplotlib.pyplot as plt
cal.plot(kind = 'bar',figsize = (10,5),color = 'skyblue')
plt.title('top 10 according to country per cases:',fontsize = '12',color = 'blue')
plt.xlabel('Country',color ='blue',fontsize = '12')
plt.ylabel('Cases',color = 'blue',fontsize = '12')
plt.grid(axis = 'y',linestyle='--',alpha = 0.7)
plt.tight_layout()
plt.legend()
plt.show()
```

```python
#Countries by Total Deaths
#Calculation: Sum Deaths per Country, sort descending, take top 10.
#cal ---------------------------------------------------------------
```

```python
death = df[df['Status'] == 'Deaths']
cal = death.groupby('Country')['Cases'].sum().sort_values(ascending = False).head(10)
cal
```

```python
#visulization -------------------------------------------------
```

```python
cal.plot(kind = 'line',color = 'green',linestyle ='--',marker = '*',markerfacecolor = 'blue',markeredgecolor = 'blue')
plt.title('Sum Deaths per Country:',color='red',fontsize = '12')
plt.xlabel('country',fontsize = '12',color = 'red')
plt.ylabel('deaths',color = 'red',fontsize = '12')
plt.legend()
plt.tight_layout()
plt.show()
```

```python
#Days with Highest New Cases Globally
#Calculation: Sum New Cases per Date, sort descending, take top 10.
#Visualization: Line or Bar chart.
```

```python
#cal-------------------------------------
```

```python
new_cases = df.groupby(['Country','Date'])['Cases'].diff().fillna(0).sort_values(ascending = False).head(10)
new_cases
```

```python
new_cases.plot(kind='bar',color = 'brown',figsize = (10,5))
plt.title('new_cases according to dates',fontsize = '12')
plt.xlabel('dates',color = 'red',fontsize = '14')
plt.ylabel('cases',color = 'red',fontsize = '14')
plt.legend()
plt.tight_layout()

plt.show()
```

```python
#recover rate -------------------------
```

```python
#cal---------------------------------
```

```python
recover = df[df['Status'] == 'Recovered']
total_recover = recover.groupby('Country')['Cases'].sum().sort_values(ascending = False)
print(total_recover)
print('----------------------------------------------------------------------------------------')
total_recovers = df.groupby('Country')['Cases'].sum().sort_values(ascending = False)
print(total_recovers)
print('----------------------------------------------------------------------------------------')
total_recover_rates = (total_recover / total_recovers * 100).round(2)
total_recover_rates
```

```python
import matplotlib.pyplot as plt
total_recover_rates.plot(kind = 'pie', cmap='Pastel1',startangle=100,autopct='%1.1f%%')
plt.title('recover rate of all countrys.',fontsize = '14',color = 'green')
plt.ylabel('')
plt.tight_layout()
plt.legend()
plt.show()
```

```python
#Top 10 Countries by Total Death------------------------------------------------------------
```

```python
#calculation -------------------------------------------------------------------------------
```

```python
death = df[df['Status'] == 'Deaths']
total_death = death.groupby('Country')['Cases'].sum()
total_death
print(total_death)
print('----------------------------------------------------------------------------------------')

total_cases = df.groupby('Country')['Cases'].sum()
print(f'total_cases_for_every_county:',total_cases)
print('-------------------------------------------------------------------------------------------')

total_death_rates = (total_death / total_cases * 100).head(10).sort_values(ascending = False).round(2)
print(total_death_rates)
```

```python
#visualization for total_death_reate ----------------------------------------------------------------------
```

```python
import matplotlib.pyplot as plt
total_death_rates.plot(kind = 'pie',figsize = (10,10),cmap='Set3',autopct = '%1.1f%%',startangle = 90)
plt.title('top 10 death reates in each country:',color = 'red',fontsize = '15')
plt.ylabel('')
plt.tight_layout()
plt.legend()
plt.show()
```

```python
#total_recover_rates
#total_death_rates
```

```python
import matplotlib.pyplot as plt

fig, ax1 = plt.subplots(figsize=(10,5))

# Bar chart for total deaths
ax1.bar(total_death_rates.index, total_death_rates.values, color='skyblue', label='Deaths')
ax1.set_xlabel('Country')
ax1.set_ylabel('Total Deaths', color='blue')
ax1.tick_params(axis='y', labelcolor='blue')

# Line chart for recovery rates on second y-axis
ax2 = ax1.twinx()
ax2.plot(total_recover_rates.index, total_recover_rates.values, color='green', marker='o', linestyle='--', label='Recovery Rate (%)')
ax2.set_ylabel('Recovery Rate (%)', color='green')
ax2.tick_params(axis='y', labelcolor='green')

# Title and layout
plt.title('Total Deaths and Recovery Rate per Country')
fig.tight_layout()
plt.show()
```
