```python
import pandas as pd
df = pd.read_csv(r"C:\Users\indur\Downloads\messy_covid19_data.csv")
print(df.to_string())
```

```python
#cleaning country column------------------------------------
```

```python
import pandas as pd
df['Country'] = df['Country'].fillna('unknown')
df['Country'] = df['Country'].str.replace('Inda' , 'india')
df['Country'] = df['Country'].str.replace('USA','U S A')
df['Country'] = df['Country'].str.replace('Usa','U S A')
df['Country'] = df['Country'].str.replace('Uk','U K')
df['Country']  =df['Country'].str.strip().str.title()
print(df.to_string())
```

```python
#cleaning Status column --------------------------------
import pandas as pd
df['Status'] = df['Status'].str.replace('Recoverd','Recovered')
df['Status'] = df['Status'].str.strip().str.title()
df['Status'] = df['Status'].fillna(method = 'ffill')
print(df.to_string())
```

```python
#cleaning Cases colums ------------------------------
```

```python
df['Cases'] = pd.to_numeric(df['Cases'],errors = 'coerce')
mean_Cases =  df['Cases'].mean().astype(int)
df['Cases'] = df['Cases'].fillna(mean_Cases)
df['Cases'] = df['Cases'].astype(str).str.replace('.0','',regex=False)
print(df.to_string())
```

```python
#cleaning Tests columns-----------------------------
```

```python
import pandas as pd
df['Tests'] = pd.to_numeric(df['Tests'],errors = 'coerce')
mean_Tests = df['Tests'].mean().astype(int)
df['Tests'] = df['Tests'].fillna(mean_Tests)
df['Tests'] = df['Tests'].astype(str).str.replace('.0','',regex = False)
print(df.to_string())
```

```python
#cleaning Vaccinated cloumns -------------------------------------------
```

```python
df['Vaccinated'] = pd.to_numeric(df['Vaccinated'],errors = 'coerce')
mean_Vaccinated = df['Vaccinated'].mean().astype(int)
df['Vaccinated'] = df['Vaccinated'].fillna(mean_Vaccinated)
df['Vaccinated'] = df['Vaccinated'].astype(str).str.replace('.0','',regex = False)
print(df.to_string())
```

```python
df.to_csv('cleaned_covid19_data.csv', index=False)  #  saving cleaning data set ------------------------------
```
